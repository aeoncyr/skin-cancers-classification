# Skin Cancer Model Training

This notebook demonstrates how to train the skin cancer detection model using the `skin_cancer_detection` package.

```python
import sys
import os

# Add the src directory to the path so we can import the package
sys.path.append(os.path.abspath(os.path.join(os.getcwd(), '..', 'src')))

from skin_cancer_detection import train, config, visualization
```

## 1. Configuration
Review the training parameters.

```python
print(f"Image Size: {config.IMG_SIZE}")
print(f"Batch Size: {config.BATCH_SIZE}")
print(f"Epochs: {config.EPOCHS}")
print(f"Learning Rate: {config.LEARNING_RATE}")
```

## 2. Train Model
Start the training process. This will load data, build the model, and execute the training loop.

```python
# You can override parameters here
history, model = train.train(
    epochs=5, # Reduced for demo purposes, use config.EPOCHS or more for real training
    learning_rate=0.001,
    save_path='../skin_cancer_cnn_model_notebook.h5'
)
```

## 3. Visualize Results
Plot the accuracy and loss curves.

```python
if history:
    visualization.plot_training_history(history)
```
