# Skin Cancer Data Understanding

This notebook demonstrates how to explore the HAM10000 dataset using the `skin_cancer_detection` package.

```python
import sys
import os

# Add the src directory to the path so we can import the package
sys.path.append(os.path.abspath(os.path.join(os.getcwd(), '..', 'src')))

from skin_cancer_detection import data, visualization, config
import pandas as pd
```

Understanding the balance of the dataset classes.
